WIN RATE 

```python
win=df.groupby('winner').size()
win.sort_values(ascending=False)
```

```python
total_matches=pd.concat([df['team1'],df['team2']]).value_counts()
total_matches
```

```python
win_rate=(win/total_matches)*100
win_rate.sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
```

```python
z=df['team1'].unique()
z
```

```python
win_rate.plot(kind="bar")
plt.title("Win rate")
```

```python
plt.savefig('win_rate.png', bbox_inches='tight')
```

IMPACT OF TOSS


```python
#won toss and won match
z=df.loc[df['toss_winner']==df['winner']]
len(z)
```

```python
fig, ax = plt.subplots()
labels = 'Won', 'Lost',
sizes = [len(z),len(df)-len(z)]
ax.pie(sizes, labels=labels)
plt.title("impact of toss wins")
```

```python
plt.savefig('toss_impact', bbox_inches='tight')
```

```python
#chaseing team won
chasing=df.loc[df['win_by_wickets'] !=0]
chasing
```

CHASING BIAS

```python
fig, ax = plt.subplots()
labels = 'Won', 'Lost',
sizes = [len(chasing),len(df)-len(chasing)]
ax.pie(sizes, labels=labels)
plt.title("chasing team")
```

```python
plt.savefig('chasing_bias', bbox_inches='tight')
```
